### 43. Ek simple calculator banayein (+, -, *, /) symbols ke basis par operation perform karein.

```python
num1 = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if op == '+':
    print(num1 + num2)
elif op == '-':
    print(num1 - num2)
elif op == '*':
    print(num1 * num2)
elif op == '/':
    if num2 != 0:
        print(num1 / num2)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

### 44. User ka weight aur height le kar BMI category batayein (Underweight, Normal, Overweight).

```python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))
bmi = weight / (height ** 2)

if bmi < 18.5:
    print("Underweight")
elif bmi < 24.9:
    print("Normal")
elif bmi < 29.9:
    print("Overweight")
else:
    print("Obese")
```

### 45. ATM Machine Logic: Pehle PIN check karein, agar sahi ho toh Amount poochiye, agar account balance se kam amount ho toh 'Withdrawal Success'.

```python
correct_pin = "1234"
balance = 10000

pin = input("Enter PIN: ")
if pin == correct_pin:
    amount = float(input("Enter withdrawal amount: "))
    if amount <= balance:
        balance -= amount
        print("Withdrawal Success. Remaining balance:", balance)
    else:
        print("Insufficient balance")
else:
    print("Wrong PIN")
```

### 58. Electricity Bill: Pehle 100 units free, next 100 units Rs 5/unit, uske baad Rs 10/unit.

```python
units = float(input("Enter units consumed: "))
if units <= 100:
    bill = 0
elif units <= 200:
    bill = (units - 100) * 5
else:
    bill = (100 * 5) + (units - 200) * 10
print("Electricity Bill: Rs", bill)
```

### 60. Final Challenge: Ek mini-project logic banayein jisme user se Username, Password, aur ek Security Question poochein nested loops aur if-else ke saath.

```python
username = input("Set username: ")
password = input("Set password: ")
security_ans = input("Set security answer: ")

print("\n--- Login ---")
u = input("Enter username: ")
if u == username:
    p = input("Enter password: ")
    if p == password:
        print("Login Successful")
    else:
        print("Wrong password. Security question:")
        ans = input("What is your security answer? ")
        if ans == security_ans:
            print("Security answer correct. Reset your password.")
        else:
            print("Access Denied")
else:
    print("Username not found")
```
